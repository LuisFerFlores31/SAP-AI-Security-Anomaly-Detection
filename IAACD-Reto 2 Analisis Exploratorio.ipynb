```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
pd.set_option("display.max_columns", 100)
```

```python
#direccion del csv
df = pd.read_csv(DATA_PATH)

df_original = df.copy(deep=True)
df.columns = df.columns.astype(str).str.strip()

print(f"Filas: {df.shape[0]:,}")
print(f"Columnas: {df.shape[1]:,}")

display(df.head())
df.info()

print("\nNombres completos de las columnas:")
print(df.columns.tolist())

display(df.describe(include="all").T)
```

```python
columnas_fecha = [
    c for c in df.columns
    if c == "@timestamp" or c.startswith("@event_time")
]

#metricas potenciales, para luego ver cuales son numericas
candidatas_metricas = [
    c for c in df.columns
    if c.startswith("sap_llm_response")
    or (
        c.startswith("llm_")
        and any(p in c.lower() for p in
                ["token", "latency", "duration", "cost"])
    )
]

def clasificar_variable(c):
    nombre = c.lower()

    if c in columnas_fecha:
        return "Temporal"
    if c == "_score":
        return "Numerica: metadato de busqueda"
    if "status" in nombre or "version" in nombre:
        return "Categórica: código o versión"
    if (
        nombre.endswith("_id")
        or nombre.endswith("_hash")
        or c in ["_id", "client_ip"]
    ):
        return "Identificador"
    if c in candidatas_metricas:
        return "Métrica numérica potencial: verificar significado"
    if pd.api.types.is_numeric_dtype(df[c]):
        return "Numérica: verificar significado"
    return "Categórica o texto"

tipos = pd.DataFrame({
    "variable": df.columns,
    "tipo_pandas": [str(df[c].dtype) for c in df.columns],
    "clasificacion": [clasificar_variable(c) for c in df.columns],
    "valores_distintos": [df[c].nunique() for c in df.columns],
    "porcentaje_nulos": [df[c].isna().mean() * 100 for c in df.columns]
})

display(tipos.round(2))
```

```python
#unificar celdas vacias o compuestas solo por espacios
df = df.replace(r"^\s*$", np.nan, regex=True)

#convertir fechas y medimos los valores que no pudieron interpretarse
for c in columnas_fecha:
    original = df[c].copy()
    convertido = pd.to_datetime(
        original, errors="coerce", utc=True, format="mixed"
    )

    fallos = (original.notna() & convertido.isna()).sum()
    print(f"{c}: {fallos} valores no se pudieron convertir a fecha.")

    df[c] = convertido

#solo convertir una metrica si todos sus valores presentes
for c in candidatas_metricas:
    convertido = pd.to_numeric(df[c], errors="coerce")
    fallos = (df[c].notna() & convertido.isna()).sum()

    if fallos == 0:
        df[c] = convertido
    else:
        print(f"Revisar {c}: {fallos} valores no numéricos; se conserva.")

#reportar duplicados sin borrarlos todavia
print("Filas exactamente duplicadas:", df.duplicated().sum())

if "_id" in df.columns:
    ids_presentes = df["_id"].dropna()
    print("Repeticiones adicionales de _id:", ids_presentes.duplicated().sum())

#activar solamente si son duplicados de extraccion
ELIMINAR_DUPLICADOS = False
if ELIMINAR_DUPLICADOS:
    df = df.drop_duplicates().copy()

faltantes = pd.DataFrame({
    "cantidad": df.isna().sum(),
    "porcentaje": df.isna().mean() * 100
}).sort_values("porcentaje", ascending=False)

display(faltantes.round(2))

con_nulos = faltantes[faltantes["cantidad"] > 0].head(25)

if not con_nulos.empty:
    con_nulos["porcentaje"].sort_values().plot.barh(
        figsize=(10, 7), color="steelblue"
    )
    plt.title("Columnas con mayor porcentaje de valores faltantes")
    plt.xlabel("Valores faltantes (%)")
    plt.ylabel("Columna")
    plt.tight_layout()
    plt.show()

#metricas para los siguientes apartados
metricas = [
    c for c in candidatas_metricas
    if pd.api.types.is_numeric_dtype(df[c])
    and df[c].notna().any()
]

print("Métricas numéricas disponibles:", metricas)
```

```python
COL_TIPO = "sap_function_log_type"

if COL_TIPO not in df.columns:
    raise KeyError(f"No existe la columna {COL_TIPO}.")

print(f"Registros: {len(df):,}")
print(f"Variables: {df.shape[1]:,}")

tipos_evento = (
    df[COL_TIPO]
    .astype("string")
    .str.strip()
    .fillna("(Sin dato)")
    .value_counts()
)

resumen_tipos = tipos_evento.to_frame("registros")
resumen_tipos["porcentaje"] = (
    resumen_tipos["registros"] / len(df) * 100
)

display(resumen_tipos.round(2))

tipos_evento.head(15).sort_values().plot.barh(
    figsize=(10, 4), color="steelblue"
)
plt.title("Distribucion general de tipos de evento")
plt.xlabel("Cantidad de registros")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
fecha_principal = (
    "@timestamp" if "@timestamp" in columnas_fecha
    else next(iter(columnas_fecha), None)
)

if fecha_principal:
    eventos_diarios = (
        df.dropna(subset=[fecha_principal])
          .set_index(fecha_principal)
          .resample("D")
          .size()
    )

    if not eventos_diarios.empty:
        eventos_diarios.plot(marker="o", color="darkorange")
        plt.title("Registros por día")
        plt.xlabel("Fecha (UTC)")
        plt.ylabel("Cantidad de registros")
        plt.tight_layout()
        plt.show()
```

```python
#subconjunto de security
es_security = (
    df[COL_TIPO].astype("string").str.strip().str.casefold()
    .eq("security").fillna(False)
)

df_security = df.loc[es_security].copy()

if df_security.empty:
    raise ValueError(
        "No se encontraron eventos con categoria 'Security'. "
        "Revisa los valores mostrados en la celda anterior."
    )

print(f"Eventos Security: {len(df_security):,}")
print(f"Proporción del total: {len(df_security) / len(df):.2%}")

display(df_security.head())

#para diferenciar security
perfil_security = pd.DataFrame({
    "tipo": df_security.dtypes.astype(str),
    "nulos_pct": df_security.isna().mean() * 100,
    "valores_distintos": df_security.nunique(dropna=True)
}).sort_values("nulos_pct")

display(perfil_security.round(2))
```

```python
columnas_contexto = [
    c for c in ["sap_app_env", "macro_region", "service_id"]
    if c in df.columns
]

for c in columnas_contexto:
    contexto = pd.DataFrame({
        c: df[c].astype("string").fillna("(Sin dato)"),
        "es_security": es_security.astype(int)
    })

    resumen = (
        contexto.groupby(c)["es_security"]
        .agg(total="size", security="sum")
        .sort_values("security", ascending=False)
    )
    resumen["security_pct"] = resumen["security"] / resumen["total"] * 100

    display(resumen.head(15).round(2))

    resumen.head(15)[["total", "security"]].plot.bar(
        figsize=(11, 4), color=["lightgray", "teal"]
    )
    plt.title(f"Eventos totales y Security por {c}")
    plt.ylabel("Cantidad de registros")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
def evaluar_variable(c):
    s = df_security[c]
    nombre = c.lower()
    distintos = s.nunique(dropna=True)
    presentes = s.notna().sum()

    if presentes == 0:
        return "Excluir: vacia en Security"
    if distintos <= 1:
        return "Excluir como predictor: constante en Security"
    if pd.api.types.is_datetime64_any_dtype(s):
        return "Derivar: hora, dia y bloques temporales"
    if nombre in ["_score", "_index"]:
        return "Revisar: metadato de extracción"
    if (
        nombre.endswith("_id")
        or nombre.endswith("_hash")
        or nombre in ["_id", "client_ip"]
    ):
        return "Usar para agrupacion/conteos, evitar como numero"
    if pd.api.types.is_numeric_dtype(s):
        return "Candidata numerica: confirmar significado"
    if distintos / presentes > 0.9:
        return "Revisar: texto libre o alta cardinalidad"
    return "Candidata categorica"

revision_variables = perfil_security.copy()
revision_variables["uso_propuesto"] = [
    evaluar_variable(c) for c in revision_variables.index
]

display(revision_variables)
```

## Analisis integrado de errores y rendimiento

Se incorporan las visualizaciones y metricas del notebook `Explore`: errores HTTP, errores de SAP y LLM, distribucion por aplicacion, region y evolucion temporal.


```python
# Clasificaci?n de errores HTTP, SAP y LLM
df["http_status_code_num"] = pd.to_numeric(df["http_status_code"], errors="coerce")
df["is_http_error"] = df["http_status_code_num"] >= 400
df["is_http_4xx"] = df["http_status_code_num"].between(400, 499, inclusive="both")
df["is_http_5xx"] = df["http_status_code_num"] >= 500

log_type_text = df[COL_TIPO].astype("string").fillna("").str.upper()
df["is_log_error"] = log_type_text.str.contains(
    r"ERROR|TIMEOUT|CRITICAL|FAIL", regex=True, na=False
)

llm_status_text = df["llm_status"].astype("string").str.lower()
df["is_llm_error"] = (
    (df["llm_status"].notna() & llm_status_text.ne("success"))
    | df["llm_error_message"].notna()
        & df["llm_error_message"].astype("string").str.strip().ne("")
)
df["is_any_error"] = (
    df["is_http_error"] | df["is_log_error"] | df["is_llm_error"]
)

total_logs = len(df)
total_errors = int(df["is_any_error"].sum())
error_pct = total_errors / total_logs * 100 if total_logs else 0
print("=" * 60)
print("RESUMEN GENERAL DE EVENTOS Y ERRORES")
print("=" * 60)
print(f"Total de registros:          {total_logs:,}")
print(f"Total de eventos con error:  {total_errors:,} ({error_pct:.2f}%)")
print(f" - Errores HTTP (4xx / 5xx): {int(df['is_http_error'].sum()):,}")
print(f"   * Cliente (4xx):          {int(df['is_http_4xx'].sum()):,}")
print(f"   * Servidor (5xx):         {int(df['is_http_5xx'].sum()):,}")
print(f" - Errores de Log SAP:       {int(df['is_log_error'].sum()):,}")
print(f" - Errores LLM:              {int(df['is_llm_error'].sum()):,}")
```

```python
# Frecuencia de tipos de log
log_type_counts = (
    df[COL_TIPO].astype("string").fillna("(Sin dato)")
      .value_counts(dropna=False)
      .rename_axis("Log Type")
      .reset_index(name="Frecuencia")
)
log_type_counts["Porcentaje (%)"] = (
    log_type_counts["Frecuencia"] / len(df) * 100
).round(2)
display(log_type_counts)

colors = [
    "#d9534f" if any(k in str(value).upper() for k in ["ERROR", "TIMEOUT", "FAIL"])
    else "#0275d8"
    for value in log_type_counts["Log Type"]
]
plt.figure(figsize=(10, 5))
sns.barplot(data=log_type_counts, x="Log Type", y="Frecuencia", palette=colors)
plt.title("Frecuencia por tipo de log (rojo = error / timeout)", fontweight="bold")
plt.xlabel("Tipo de log")
plt.ylabel("Cantidad de eventos")
plt.xticks(rotation=45, ha="right")
for index, row in log_type_counts.iterrows():
    plt.text(index, row["Frecuencia"], f"{row['Porcentaje (%)']}%",
             ha="center", va="bottom", fontsize=9)
plt.tight_layout()
plt.show()

# Estados y mensajes de error de LLM
llm_logs = df[df["llm_status"].notna() | df["llm_model_id"].notna()]
print(f"Total de llamadas a LLM registradas: {len(llm_logs):,}")
llm_status_counts = (
    llm_logs["llm_status"].astype("string").fillna("(Sin dato)")
      .value_counts()
      .rename_axis("LLM Status")
      .reset_index(name="Frecuencia")
)
if not llm_status_counts.empty:
    llm_status_counts["Porcentaje (%)"] = (
        llm_status_counts["Frecuencia"] / len(llm_logs) * 100
    ).round(2)
    display(llm_status_counts)

llm_err_counts = (
    df["llm_error_message"].dropna().astype(str).value_counts().head(15)
      .rename_axis("Mensaje de Error LLM")
      .reset_index(name="Frecuencia")
)
if not llm_err_counts.empty:
    llm_err_counts["Porcentaje s/ errores LLM (%)"] = (
        llm_err_counts["Frecuencia"] / max(1, int(df["is_llm_error"].sum())) * 100
    ).round(2)
    display(llm_err_counts)
    plt.figure(figsize=(12, 6))
    sns.barplot(data=llm_err_counts.head(8), y="Mensaje de Error LLM",
                x="Frecuencia", color="#e74c3c")
    plt.title("Top de errores mas frecuentes en llamadas LLM", fontweight="bold")
    plt.xlabel("Frecuencia")
    plt.ylabel("Mensaje de error")
    plt.tight_layout()
    plt.show()
```

```python
# Errores por modelo, aplicaci?n y macro regi?n
if df["llm_model_id"].notna().any():
    model_errors = (
        df[df["llm_model_id"].notna()]
        .groupby("llm_model_id")
        .agg(Total_Peticiones=("is_llm_error", "count"),
             Total_Errores=("is_llm_error", "sum"))
        .reset_index()
    )
    model_errors["Tasa de Error (%)"] = (
        model_errors["Total_Errores"] / model_errors["Total_Peticiones"] * 100
    ).round(2)
    model_errors = model_errors.sort_values("Total_Errores", ascending=False)
    display(model_errors)
    plt.figure(figsize=(10, 4))
    sns.barplot(data=model_errors, x="llm_model_id", y="Tasa de Error (%)",
                palette="Reds_r")
    plt.title("Tasa de error por modelo LLM", fontweight="bold")
    plt.xlabel("Modelo")
    plt.ylabel("Tasa de error (%)")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.show()

app_errors = (
    df.groupby("sap_function_application")
      .agg(Total_Peticiones=("is_any_error", "count"),
           Errores_HTTP=("is_http_error", "sum"),
           Errores_LLM=("is_llm_error", "sum"),
           Total_Errores=("is_any_error", "sum"))
      .reset_index()
)
app_errors["Tasa de Error (%)"] = (
    app_errors["Total_Errores"] / app_errors["Total_Peticiones"] * 100
).round(2)
app_errors = app_errors.sort_values("Total_Errores", ascending=False)
display(app_errors)
plt.figure(figsize=(12, 5))
sns.barplot(data=app_errors, x="sap_function_application",
            y="Total_Errores", palette="viridis")
plt.title("Total de errores por aplicaci?n SAP", fontweight="bold")
plt.xlabel("Aplicaci?n")
plt.ylabel("Cantidad de errores")
plt.xticks(rotation=40, ha="right")
plt.tight_layout()
plt.show()

region_errors = (
    df.groupby("macro_region")
      .agg(Total_Peticiones=("is_any_error", "count"),
           Total_Errores=("is_any_error", "sum"))
      .reset_index()
)
region_errors["Tasa de Error (%)"] = (
    region_errors["Total_Errores"] / region_errors["Total_Peticiones"] * 100
).round(2)
region_errors = region_errors.sort_values("Total_Errores", ascending=False)
display(region_errors)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(data=region_errors, x="macro_region", y="Total_Errores",
            ax=axes[0], palette="flare")
axes[0].set_title("Volumen absoluto de errores por regi?n")
axes[0].tick_params(axis="x", rotation=30)
sns.barplot(data=region_errors, x="macro_region", y="Tasa de Error (%)",
            ax=axes[1], palette="crest")
axes[1].set_title("Tasa relativa de error por regi?n")
axes[1].tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Eventos Security y señales para priorizar la investigaci?n
security_mask = df[COL_TIPO].fillna("").astype("string").str.contains(
    r"\bSECURITY\b", case=False, regex=True, na=False
)
df_sec = df.loc[security_mask].copy()

if df_sec.empty:
    print("No se encontraron eventos cuyo tipo contenga SECURITY.")
else:
    df_sec["http_status_num"] = pd.to_numeric(
        df_sec["http_status_code"], errors="coerce"
    )
    message_text = df_sec["sap_function_message"].fillna("").astype(str)
    suspicious_text = message_text.str.contains(
        r"login|auth|unauthori[sz]ed|forbidden|access denied|failed|blocked|"
        r"token|password|injection|brute.?force|anomal|suspicious|attack|exploit",
        case=False, regex=True, na=False
    )
    df_sec["is_suspicious_signal"] = (
        df_sec["http_status_num"].isin([401, 403, 429])
        | (df_sec["http_status_num"] >= 500)
        | suspicious_text
    )
    print("ANaLISIS DE EVENTOS SECURITY")
    print(f"Total de registros: {len(df):,}")
    print(f"Eventos SECURITY: {len(df_sec):,} ({len(df_sec) / len(df) * 100:.2f}%)")
    print(f"Eventos con señales para investigar: "
          f"{int(df_sec['is_suspicious_signal'].sum()):,}")

    print("\nTipos de log encontrados:")
    display(df_sec[COL_TIPO].value_counts(dropna=False).to_frame("Frecuencia"))
    print("\nDistribución por código HTTP:")
    display(df_sec["http_status_num"].value_counts(dropna=False).head(15)
            .to_frame("Frecuencia"))
    print("\nAplicaciones con m?s eventos SECURITY:")
    display(df_sec["sap_function_application"].value_counts().head(10)
            .to_frame("Frecuencia"))

    print("\nIPs con m?s eventos SECURITY:")
    if "client_ip" in df_sec.columns:
        display(df_sec["client_ip"].dropna().astype(str).value_counts().head(15)
                .to_frame("Frecuencia"))
    else:
        print("La columna client_ip no esta disponible.")

    print("\nMensajes SECURITY m?s frecuentes:")
    display(df_sec["sap_function_message"].dropna().astype(str).value_counts()
            .head(15).to_frame("Frecuencia"))

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    top_apps = df_sec["sap_function_application"].value_counts().head(10)
    sns.barplot(x=top_apps.values, y=top_apps.index, ax=axes[0], color="crimson")
    axes[0].set_title("Aplicaciones afectadas")
    axes[0].set_xlabel("Eventos")

    top_regions = df_sec["macro_region"].value_counts().head(10)
    sns.barplot(x=top_regions.values, y=top_regions.index,
                ax=axes[1], color="darkorange")
    axes[1].set_title("Macro regiones")
    axes[1].set_xlabel("Eventos")

    risk_counts = (
        df_sec["is_suspicious_signal"]
        .map({False: "Sin señal", True: "Con señal"})
        .value_counts()
    )
    sns.barplot(x=risk_counts.values, y=risk_counts.index,
                ax=axes[2], color="purple")
    axes[2].set_title("Señales para investigar")
    axes[2].set_xlabel("Eventos")
    plt.tight_layout()
    plt.show()

    columnas_revision = [
        c for c in ["client_ip", "http_status_num",
                    "headers_http_request_method", "heathers_request_path",
                    "sap_function_message"]
        if c in df_sec.columns
    ]
    eventos_prioritarios = (
        df_sec.loc[df_sec["is_suspicious_signal"], columnas_revision]
        .head(25)
    )
    print("\nMuestra de eventos con señales para investigar:")
    display(eventos_prioritarios)
```

# 5. Análisis por Ventanas Temporales Solapadas (Sliding Windows: 5 min, paso: 2.5 min)

### Justificación Técnica y Metodología:
- **Duración de la ventana ($W$):** 5 minutos (`300s`). Permite capturar eventos continuos, ráfagas cortas de peticiones y correlaciones a corto plazo.
- **Paso / Desplazamiento ($S$):** 2.5 minutos (`150s`). Genera un **solapamiento del 50%** entre ventanas consecutivas:
  - Ventana 0: `[00:00:00 - 00:05:00)`
  - Ventana 1: `[00:02:30 - 00:07:30)`
  - Ventana 2: `[00:05:00 - 00:10:00)`
- **Ventaja del solapamiento:** Evita el *efecto de borde* (*boundary effect*), donde un ataque o comportamiento anómalo que ocurre entre dos ventanas fijas queda partido a la mitad y pasa desapercibido sin superar los umbrales de alerta.

```python
# 5.1 Agregación Global del Sistema por Ventanas Deslizantes (5 min con paso de 2.5 min)

# 1. Asegurar banderas necesarias
if "is_http_4xx" not in df.columns:
    http_num = pd.to_numeric(df["http_status_code"], errors="coerce")
    df["is_http_4xx"] = http_num.between(400, 499, inclusive="both")
    df["is_http_5xx"] = http_num >= 500
    df["is_http_error"] = http_num >= 400

if "is_security" not in df.columns:
    df["is_security"] = df[COL_TIPO].fillna("").astype(str).str.contains(r"\bSECURITY\b", case=False, regex=True)

if "is_any_error" not in df.columns:
    df["is_any_error"] = df["is_http_error"] | df.get("is_log_error", False) | df.get("is_llm_error", False)

# 2. Paso base de agregación temporal: 150 segundos (2.5 minutos)
col_fecha = fecha_principal if (fecha_principal and fecha_principal in df.columns) else "@timestamp"
df_indexed = df.dropna(subset=[col_fecha]).set_index(col_fecha).sort_index()

STEP = "150s"      # 2.5 minutos
ROLLING_STEPS = 2  # 2 pasos de 2.5 min = 5 minutos de ventana

# Resumen base por subintervalo de 2.5 min (optimizado vectorialmente)
df_subintervals = df_indexed.resample(STEP).agg(
    total_events=("event_hash", "count"),
    http_4xx=("is_http_4xx", "sum"),
    http_5xx=("is_http_5xx", "sum"),
    total_errors=("is_any_error", "sum"),
    security_events=("is_security", "sum"),
    unique_ips=("client_ip", "nunique"),
    total_tokens=("llm_total_tokens", "sum"),
    avg_latency_ms=("sap_llm_response_time", "mean")
)

# 3. Aplicar ventana deslizante de 5 minutos (rolling de 2 pasos)
df_windows_5m = pd.DataFrame(index=df_subintervals.index)

# Métricas acumuladas en la ventana de 5 minutos
df_windows_5m["total_events"] = df_subintervals["total_events"].rolling(ROLLING_STEPS).sum()
df_windows_5m["http_4xx"] = df_subintervals["http_4xx"].rolling(ROLLING_STEPS).sum()
df_windows_5m["http_5xx"] = df_subintervals["http_5xx"].rolling(ROLLING_STEPS).sum()
df_windows_5m["total_errors"] = df_subintervals["total_errors"].rolling(ROLLING_STEPS).sum()
df_windows_5m["security_events"] = df_subintervals["security_events"].rolling(ROLLING_STEPS).sum()
df_windows_5m["total_tokens"] = df_subintervals["total_tokens"].rolling(ROLLING_STEPS).sum()
df_windows_5m["avg_latency_ms"] = df_subintervals["avg_latency_ms"].rolling(ROLLING_STEPS).mean()

# Descartar la primera ventana parcial (requiere 2 pasos para completar 5 min)
df_windows_5m = df_windows_5m.dropna(subset=["total_events"]).copy()

# Definir inicio y fin explícito de cada ventana
df_windows_5m["window_end"] = df_windows_5m.index
df_windows_5m["window_start"] = df_windows_5m["window_end"] - pd.Timedelta(minutes=5)

# Métricas relativas por ventana
df_windows_5m["error_rate_%"] = (df_windows_5m["total_errors"] / df_windows_5m["total_events"] * 100).fillna(0).round(2)
df_windows_5m["security_rate_%"] = (df_windows_5m["security_events"] / df_windows_5m["total_events"] * 100).fillna(0).round(2)

print("ANÁLISIS DE VENTANAS TEMPORALES SOLAPADAS (5 MIN, PASO 2.5 MIN)")
print("=" * 70)
print(f"Total de ventanas generadas:    {len(df_windows_5m):,}")
print(f"Rango temporal cubierto:        {df_windows_5m['window_start'].min()} a {df_windows_5m['window_end'].max()}")
print(f"Eventos promedio por ventana:   {df_windows_5m['total_events'].mean():.2f} (std: {df_windows_5m['total_events'].std():.2f})")
print(f"Errores promedio por ventana:   {df_windows_5m['total_errors'].mean():.2f}")
print(f"Security eventos prom/ventana:  {df_windows_5m['security_events'].mean():.2f}")
print(f"Tasa de error global promedio:  {df_windows_5m['error_rate_%'].mean():.2f}%")

display(df_windows_5m[["window_start", "window_end", "total_events", "total_errors", "error_rate_%", "security_events", "total_tokens"]].head(6))
```

```python
# 5.2 Visualización de Tráfico, Errores y Seguridad en Ventanas de 5 min

fig, axes = plt.subplots(3, 1, figsize=(15, 11), sharex=True)

# 1. Tráfico total por ventana con umbrales de media y desviación
media_trafico = df_windows_5m["total_events"].mean()
std_trafico = df_windows_5m["total_events"].std()
umbral_alto_trafico = media_trafico + 3 * std_trafico

axes[0].plot(df_windows_5m["window_end"], df_windows_5m["total_events"], color="#1f77b4", linewidth=1.2, label="Peticiones / 5 min")
axes[0].axhline(media_trafico, color="navy", linestyle="--", alpha=0.8, label=f"Media ({media_trafico:.0f})")
axes[0].axhline(umbral_alto_trafico, color="crimson", linestyle=":", linewidth=1.5, label=f"Umbral Anómalo (+3σ: {umbral_alto_trafico:.0f})")
axes[0].fill_between(df_windows_5m["window_end"], media_trafico - 2*std_trafico, media_trafico + 2*std_trafico, color="#1f77b4", alpha=0.1, label="Banda normal (±2σ)")
axes[0].set_title("Volumen de Tráfico Global por Ventanas de 5 min (Paso: 2.5 min)", fontweight="bold")
axes[0].set_ylabel("Total Eventos")
axes[0].legend(loc="upper right")

# 2. Errores totales y desglosados (4xx vs 5xx)
axes[1].plot(df_windows_5m["window_end"], df_windows_5m["total_errors"], color="#d62728", linewidth=1.2, label="Total Errores")
axes[1].plot(df_windows_5m["window_end"], df_windows_5m["http_4xx"], color="#ff7f0e", alpha=0.7, linewidth=1, label="Errores 4xx (Cliente)")
axes[1].plot(df_windows_5m["window_end"], df_windows_5m["http_5xx"], color="#9467bd", alpha=0.7, linewidth=1, label="Errores 5xx (Servidor)")
axes[1].set_title("Frecuencia de Errores por Ventana de 5 min", fontweight="bold")
axes[1].set_ylabel("Cantidad de Errores")
axes[1].legend(loc="upper right")

# 3. Eventos Security y Tasa de Error (%)
color_sec = "#2ca02c"
color_rate = "#8c564b"
axes[2].plot(df_windows_5m["window_end"], df_windows_5m["security_events"], color=color_sec, linewidth=1.2, label="Eventos SECURITY")
axes[2].set_title("Eventos de Seguridad (SECURITY Logs) por Ventana de 5 min", fontweight="bold")
axes[2].set_ylabel("Eventos SECURITY", color=color_sec)
axes[2].tick_params(axis="y", labelcolor=color_sec)

# Eje secundario para la tasa de error
ax2_twin = axes[2].twinx()
ax2_twin.plot(df_windows_5m["window_end"], df_windows_5m["error_rate_%"], color=color_rate, alpha=0.5, linestyle="--", linewidth=1, label="Tasa de Error (%)")
ax2_twin.set_ylabel("Tasa de Error (%)", color=color_rate)
ax2_twin.tick_params(axis="y", labelcolor=color_rate)
ax2_twin.grid(False)

plt.xlabel("Fecha y Hora (UTC)")
plt.tight_layout()
plt.show()
```

```python
# 5.3 Detección de Ventanas Globales Anómalas (Picos de Tráfico, Errores y Seguridad)

# Cálculo de Z-Score para identificar picos estadísticos
for col in ["total_events", "total_errors", "security_events"]:
    mean_val = df_windows_5m[col].mean()
    std_val = df_windows_5m[col].std()
    df_windows_5m[f"zscore_{col}"] = (df_windows_5m[col] - mean_val) / std_val

# Criterio de anomalía: Z-Score >= 3.0 en cualquiera de los 3 ejes
df_windows_5m["es_anomalia_trafico"] = df_windows_5m["zscore_total_events"] >= 3.0
df_windows_5m["es_anomalia_error"] = df_windows_5m["zscore_total_errors"] >= 3.0
df_windows_5m["es_anomalia_seguridad"] = df_windows_5m["zscore_security_events"] >= 3.0
df_windows_5m["es_ventana_anomala"] = (
    df_windows_5m["es_anomalia_trafico"] | 
    df_windows_5m["es_anomalia_error"] | 
    df_windows_5m["es_anomalia_seguridad"]
)

anomalias_df = df_windows_5m[df_windows_5m["es_ventana_anomala"]].copy()

print(f"Total de ventanas analizadas: {len(df_windows_5m)}")
print(f"Ventanas anómalas detectadas (Z >= 3.0): {len(anomalias_df)} ({len(anomalias_df)/len(df_windows_5m)*100:.2f}%)")
print(f" - Anomalías por picos de tráfico:     {df_windows_5m['es_anomalia_trafico'].sum()}")
print(f" - Anomalías por picos de errores:     {df_windows_5m['es_anomalia_error'].sum()}")
print(f" - Anomalías por picos de seguridad:   {df_windows_5m['es_anomalia_seguridad'].sum()}")

# Visualizar Top 10 ventanas más críticas según el Z-Score de seguridad
print("\nTOP 10 VENTANAS CON MAYOR ACTIVIDAD ANÓMALA DE SEGURIDAD:")
cols_resumen = ["window_start", "window_end", "security_events", "zscore_security_events", "total_events", "total_errors", "error_rate_%"]
display(anomalias_df.sort_values("security_events", ascending=False)[cols_resumen].head(10).round(2))

# Visualizar Top 10 ventanas con mayor volumen de tráfico (posibles ráfagas/DDoS)
print("\nTOP 10 VENTANAS CON MAYOR VOLUMEN DE TRÁFICO (PICOS DE CARGA):")
display(anomalias_df.sort_values("total_events", ascending=False)[["window_start", "window_end", "total_events", "zscore_total_events", "total_errors", "security_events"]].head(10).round(2))
```

```python
# 5.4 Análisis de Ventanas Temporales por IP (Detección de Ráfagas y Ataques por IP)

# Agrupar por IP y por subintervalos de 2.5 min
ip_subintervals = df.groupby(["client_ip", pd.Grouper(key=col_fecha, freq=STEP)]).agg(
    ip_requests=("event_hash", "count"),
    ip_errors=("is_any_error", "sum"),
    ip_security=("is_security", "sum")
).reset_index()

# Aplicar ventana deslizante de 5 min por IP (considerando timestamps continuos)
ip_subintervals = ip_subintervals.set_index(col_fecha).sort_index()

# Rolling time-based de 5 min por IP
ip_windows_5m = ip_subintervals.groupby("client_ip").rolling("5min").sum().reset_index()
ip_windows_5m["ip_error_rate_%"] = (ip_windows_5m["ip_errors"] / ip_windows_5m["ip_requests"] * 100).fillna(0).round(1)

print("ANÁLISIS DE VENTANAS DE 5 MIN POR DIRECCIÓN IP")
print("=" * 70)
print(f"Total combinaciones IP-Ventana generadas: {len(ip_windows_5m):,}")

# Top IPs con ráfagas más altas de peticiones en una ventana de 5 min
print("\nTop 10 Ráfagas más intensas de peticiones por IP en 5 minutos:")
display(ip_windows_5m.sort_values("ip_requests", ascending=False).head(10))

# Top IPs con mayor número de errores concentrados en 5 minutos
print("\nTop 10 Concentraciones de errores por IP en 5 minutos:")
display(ip_windows_5m.sort_values("ip_errors", ascending=False).head(10))

# Top IPs con mayor número de eventos de seguridad en 5 minutos
print("\nTop 10 Concentraciones de eventos SECURITY por IP en 5 minutos:")
display(ip_windows_5m.sort_values("ip_security", ascending=False).head(10))

# Gráfico de las IPs más críticas en ventanas de 5 min
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Max peticiones por IP
max_req_per_ip = ip_windows_5m.groupby("client_ip")["ip_requests"].max().nlargest(10)
sns.barplot(x=max_req_per_ip.values, y=max_req_per_ip.index, ax=axes[0], palette="Blues_r")
axes[0].set_title("Pico Máximo de Peticiones en 5 min (por IP)")
axes[0].set_xlabel("Peticiones máx en 5 min")

# Max errores por IP
max_err_per_ip = ip_windows_5m.groupby("client_ip")["ip_errors"].max().nlargest(10)
sns.barplot(x=max_err_per_ip.values, y=max_err_per_ip.index, ax=axes[1], palette="Reds_r")
axes[1].set_title("Pico Máximo de Errores en 5 min (por IP)")
axes[1].set_xlabel("Errores máx en 5 min")

# Max eventos security por IP
max_sec_per_ip = ip_windows_5m.groupby("client_ip")["ip_security"].max().nlargest(10)
sns.barplot(x=max_sec_per_ip.values, y=max_sec_per_ip.index, ax=axes[2], palette="Greens_r")
axes[2].set_title("Pico Máximo de Eventos SECURITY en 5 min (por IP)")
axes[2].set_xlabel("Eventos Security máx en 5 min")

plt.tight_layout()
plt.show()
```
