```python
#el dataset tiene que estar en la misma carpeta que el notebook
DATA_PATH = "logs/logs.csv"
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
pd.set_option("display.max_columns", 100)
```

```python
#direccion del csv
df = pd.read_csv(DATA_PATH)

df_original = df.copy(deep=True)
df.columns = df.columns.astype(str).str.strip()

print(f"Filas: {df.shape[0]:,}")
print(f"Columnas: {df.shape[1]:,}")

display(df.head())
df.info()

print("\nNombres completos de las columnas:")
print(df.columns.tolist())

display(df.describe(include="all").T)
```

```python
columnas_fecha = [
    c for c in df.columns
    if c == "@timestamp" or c.startswith("@event_time")
]

#metricas potenciales, para luego ver cuales son numericas
candidatas_metricas = [
    c for c in df.columns
    if c.startswith("sap_llm_response")
    or (
        c.startswith("llm_")
        and any(p in c.lower() for p in
                ["token", "latency", "duration", "cost"])
    )
]

def clasificar_variable(c):
    nombre = c.lower()

    if c in columnas_fecha:
        return "Temporal"
    if c == "_score":
        return "Numerica: metadato de busqueda"
    if "status" in nombre or "version" in nombre:
        return "Categórica: código o versión"
    if (
        nombre.endswith("_id")
        or nombre.endswith("_hash")
        or c in ["_id", "client_ip"]
    ):
        return "Identificador"
    if c in candidatas_metricas:
        return "Métrica numérica potencial: verificar significado"
    if pd.api.types.is_numeric_dtype(df[c]):
        return "Numérica: verificar significado"
    return "Categórica o texto"

tipos = pd.DataFrame({
    "variable": df.columns,
    "tipo_pandas": [str(df[c].dtype) for c in df.columns],
    "clasificacion": [clasificar_variable(c) for c in df.columns],
    "valores_distintos": [df[c].nunique() for c in df.columns],
    "porcentaje_nulos": [df[c].isna().mean() * 100 for c in df.columns]
})

display(tipos.round(2))
```

```python
#unificar celdas vacias o compuestas solo por espacios
df = df.replace(r"^\s*$", np.nan, regex=True)

#convertir fechas y medimos los valores que no pudieron interpretarse
for c in columnas_fecha:
    original = df[c].copy()
    convertido = pd.to_datetime(
        original, errors="coerce", utc=True, format="mixed"
    )

    fallos = (original.notna() & convertido.isna()).sum()
    print(f"{c}: {fallos} valores no se pudieron convertir a fecha.")

    df[c] = convertido

#solo convertir una metrica si todos sus valores presentes
for c in candidatas_metricas:
    convertido = pd.to_numeric(df[c], errors="coerce")
    fallos = (df[c].notna() & convertido.isna()).sum()

    if fallos == 0:
        df[c] = convertido
    else:
        print(f"Revisar {c}: {fallos} valores no numéricos; se conserva.")

#reportar duplicados sin borrarlos todavia
print("Filas exactamente duplicadas:", df.duplicated().sum())

if "_id" in df.columns:
    ids_presentes = df["_id"].dropna()
    print("Repeticiones adicionales de _id:", ids_presentes.duplicated().sum())

#activar solamente si son duplicados de extraccion
ELIMINAR_DUPLICADOS = False
if ELIMINAR_DUPLICADOS:
    df = df.drop_duplicates().copy()

faltantes = pd.DataFrame({
    "cantidad": df.isna().sum(),
    "porcentaje": df.isna().mean() * 100
}).sort_values("porcentaje", ascending=False)

display(faltantes.round(2))

con_nulos = faltantes[faltantes["cantidad"] > 0].head(25)

if not con_nulos.empty:
    con_nulos["porcentaje"].sort_values().plot.barh(
        figsize=(10, 7), color="steelblue"
    )
    plt.title("Columnas con mayor porcentaje de valores faltantes")
    plt.xlabel("Valores faltantes (%)")
    plt.ylabel("Columna")
    plt.tight_layout()
    plt.show()

#metricas para los siguientes apartados
metricas = [
    c for c in candidatas_metricas
    if pd.api.types.is_numeric_dtype(df[c])
    and df[c].notna().any()
]

print("Métricas numéricas disponibles:", metricas)
```

```python
COL_TIPO = "sap_function_log_type"

if COL_TIPO not in df.columns:
    raise KeyError(f"No existe la columna {COL_TIPO}.")

print(f"Registros: {len(df):,}")
print(f"Variables: {df.shape[1]:,}")

tipos_evento = (
    df[COL_TIPO]
    .astype("string")
    .str.strip()
    .fillna("(Sin dato)")
    .value_counts()
)

resumen_tipos = tipos_evento.to_frame("registros")
resumen_tipos["porcentaje"] = (
    resumen_tipos["registros"] / len(df) * 100
)

display(resumen_tipos.round(2))

tipos_evento.head(15).sort_values().plot.barh(
    figsize=(10, 4), color="steelblue"
)
plt.title("Distribucion general de tipos de evento")
plt.xlabel("Cantidad de registros")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
fecha_principal = (
    "@timestamp" if "@timestamp" in columnas_fecha
    else next(iter(columnas_fecha), None)
)

if fecha_principal:
    eventos_diarios = (
        df.dropna(subset=[fecha_principal])
          .set_index(fecha_principal)
          .resample("D")
          .size()
    )

    if not eventos_diarios.empty:
        eventos_diarios.plot(marker="o", color="darkorange")
        plt.title("Registros por día")
        plt.xlabel("Fecha (UTC)")
        plt.ylabel("Cantidad de registros")
        plt.tight_layout()
        plt.show()
```

```python
#subconjunto de security
es_security = (
    df[COL_TIPO].astype("string").str.strip().str.casefold()
    .eq("security").fillna(False)
)

df_security = df.loc[es_security].copy()

if df_security.empty:
    raise ValueError(
        "No se encontraron eventos con categoria 'Security'. "
        "Revisa los valores mostrados en la celda anterior."
    )

print(f"Eventos Security: {len(df_security):,}")
print(f"Proporción del total: {len(df_security) / len(df):.2%}")

display(df_security.head())

#para diferenciar security
perfil_security = pd.DataFrame({
    "tipo": df_security.dtypes.astype(str),
    "nulos_pct": df_security.isna().mean() * 100,
    "valores_distintos": df_security.nunique(dropna=True)
}).sort_values("nulos_pct")

display(perfil_security.round(2))
```

```python
columnas_contexto = [
    c for c in ["sap_app_env", "macro_region", "service_id"]
    if c in df.columns
]

for c in columnas_contexto:
    contexto = pd.DataFrame({
        c: df[c].astype("string").fillna("(Sin dato)"),
        "es_security": es_security.astype(int)
    })

    resumen = (
        contexto.groupby(c)["es_security"]
        .agg(total="size", security="sum")
        .sort_values("security", ascending=False)
    )
    resumen["security_pct"] = resumen["security"] / resumen["total"] * 100

    display(resumen.head(15).round(2))

    resumen.head(15)[["total", "security"]].plot.bar(
        figsize=(11, 4), color=["lightgray", "teal"]
    )
    plt.title(f"Eventos totales y Security por {c}")
    plt.ylabel("Cantidad de registros")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
```

```python
def evaluar_variable(c):
    s = df_security[c]
    nombre = c.lower()
    distintos = s.nunique(dropna=True)
    presentes = s.notna().sum()

    if presentes == 0:
        return "Excluir: vacia en Security"
    if distintos <= 1:
        return "Excluir como predictor: constante en Security"
    if pd.api.types.is_datetime64_any_dtype(s):
        return "Derivar: hora, dia y bloques temporales"
    if nombre in ["_score", "_index"]:
        return "Revisar: metadato de extracción"
    if (
        nombre.endswith("_id")
        or nombre.endswith("_hash")
        or nombre in ["_id", "client_ip"]
    ):
        return "Usar para agrupacion/conteos, evitar como numero"
    if pd.api.types.is_numeric_dtype(s):
        return "Candidata numerica: confirmar significado"
    if distintos / presentes > 0.9:
        return "Revisar: texto libre o alta cardinalidad"
    return "Candidata categorica"

revision_variables = perfil_security.copy()
revision_variables["uso_propuesto"] = [
    evaluar_variable(c) for c in revision_variables.index
]

display(revision_variables)
```

## Analisis integrado de errores y rendimiento

Se incorporan las visualizaciones y metricas del notebook `Explore`: errores HTTP, errores de SAP y LLM, distribucion por aplicacion, region y evolucion temporal.


```python
# Clasificaci?n de errores HTTP, SAP y LLM
df["http_status_code_num"] = pd.to_numeric(df["http_status_code"], errors="coerce")
df["is_http_error"] = df["http_status_code_num"] >= 400
df["is_http_4xx"] = df["http_status_code_num"].between(400, 499, inclusive="both")
df["is_http_5xx"] = df["http_status_code_num"] >= 500

log_type_text = df[COL_TIPO].astype("string").fillna("").str.upper()
df["is_log_error"] = log_type_text.str.contains(
    r"ERROR|TIMEOUT|CRITICAL|FAIL", regex=True, na=False
)

llm_status_text = df["llm_status"].astype("string").str.lower()
df["is_llm_error"] = (
    (df["llm_status"].notna() & llm_status_text.ne("success"))
    | df["llm_error_message"].notna()
        & df["llm_error_message"].astype("string").str.strip().ne("")
)
df["is_any_error"] = (
    df["is_http_error"] | df["is_log_error"] | df["is_llm_error"]
)

total_logs = len(df)
total_errors = int(df["is_any_error"].sum())
error_pct = total_errors / total_logs * 100 if total_logs else 0
print("=" * 60)
print("RESUMEN GENERAL DE EVENTOS Y ERRORES")
print("=" * 60)
print(f"Total de registros:          {total_logs:,}")
print(f"Total de eventos con error:  {total_errors:,} ({error_pct:.2f}%)")
print(f" - Errores HTTP (4xx / 5xx): {int(df['is_http_error'].sum()):,}")
print(f"   * Cliente (4xx):          {int(df['is_http_4xx'].sum()):,}")
print(f"   * Servidor (5xx):         {int(df['is_http_5xx'].sum()):,}")
print(f" - Errores de Log SAP:       {int(df['is_log_error'].sum()):,}")
print(f" - Errores LLM:              {int(df['is_llm_error'].sum()):,}")
```

```python
# Frecuencia de tipos de log
log_type_counts = (
    df[COL_TIPO].astype("string").fillna("(Sin dato)")
      .value_counts(dropna=False)
      .rename_axis("Log Type")
      .reset_index(name="Frecuencia")
)
log_type_counts["Porcentaje (%)"] = (
    log_type_counts["Frecuencia"] / len(df) * 100
).round(2)
display(log_type_counts)

colors = [
    "#d9534f" if any(k in str(value).upper() for k in ["ERROR", "TIMEOUT", "FAIL"])
    else "#0275d8"
    for value in log_type_counts["Log Type"]
]
plt.figure(figsize=(10, 5))
sns.barplot(data=log_type_counts, x="Log Type", y="Frecuencia", palette=colors)
plt.title("Frecuencia por tipo de log (rojo = error / timeout)", fontweight="bold")
plt.xlabel("Tipo de log")
plt.ylabel("Cantidad de eventos")
plt.xticks(rotation=45, ha="right")
for index, row in log_type_counts.iterrows():
    plt.text(index, row["Frecuencia"], f"{row['Porcentaje (%)']}%",
             ha="center", va="bottom", fontsize=9)
plt.tight_layout()
plt.show()

# Estados y mensajes de error de LLM
llm_logs = df[df["llm_status"].notna() | df["llm_model_id"].notna()]
print(f"Total de llamadas a LLM registradas: {len(llm_logs):,}")
llm_status_counts = (
    llm_logs["llm_status"].astype("string").fillna("(Sin dato)")
      .value_counts()
      .rename_axis("LLM Status")
      .reset_index(name="Frecuencia")
)
if not llm_status_counts.empty:
    llm_status_counts["Porcentaje (%)"] = (
        llm_status_counts["Frecuencia"] / len(llm_logs) * 100
    ).round(2)
    display(llm_status_counts)

llm_err_counts = (
    df["llm_error_message"].dropna().astype(str).value_counts().head(15)
      .rename_axis("Mensaje de Error LLM")
      .reset_index(name="Frecuencia")
)
if not llm_err_counts.empty:
    llm_err_counts["Porcentaje s/ errores LLM (%)"] = (
        llm_err_counts["Frecuencia"] / max(1, int(df["is_llm_error"].sum())) * 100
    ).round(2)
    display(llm_err_counts)
    plt.figure(figsize=(12, 6))
    sns.barplot(data=llm_err_counts.head(8), y="Mensaje de Error LLM",
                x="Frecuencia", color="#e74c3c")
    plt.title("Top de errores mas frecuentes en llamadas LLM", fontweight="bold")
    plt.xlabel("Frecuencia")
    plt.ylabel("Mensaje de error")
    plt.tight_layout()
    plt.show()
```

```python
# Errores por modelo, aplicaci?n y macro regi?n
if df["llm_model_id"].notna().any():
    model_errors = (
        df[df["llm_model_id"].notna()]
        .groupby("llm_model_id")
        .agg(Total_Peticiones=("is_llm_error", "count"),
             Total_Errores=("is_llm_error", "sum"))
        .reset_index()
    )
    model_errors["Tasa de Error (%)"] = (
        model_errors["Total_Errores"] / model_errors["Total_Peticiones"] * 100
    ).round(2)
    model_errors = model_errors.sort_values("Total_Errores", ascending=False)
    display(model_errors)
    plt.figure(figsize=(10, 4))
    sns.barplot(data=model_errors, x="llm_model_id", y="Tasa de Error (%)",
                palette="Reds_r")
    plt.title("Tasa de error por modelo LLM", fontweight="bold")
    plt.xlabel("Modelo")
    plt.ylabel("Tasa de error (%)")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.show()

app_errors = (
    df.groupby("sap_function_application")
      .agg(Total_Peticiones=("is_any_error", "count"),
           Errores_HTTP=("is_http_error", "sum"),
           Errores_LLM=("is_llm_error", "sum"),
           Total_Errores=("is_any_error", "sum"))
      .reset_index()
)
app_errors["Tasa de Error (%)"] = (
    app_errors["Total_Errores"] / app_errors["Total_Peticiones"] * 100
).round(2)
app_errors = app_errors.sort_values("Total_Errores", ascending=False)
display(app_errors)
plt.figure(figsize=(12, 5))
sns.barplot(data=app_errors, x="sap_function_application",
            y="Total_Errores", palette="viridis")
plt.title("Total de errores por aplicaci?n SAP", fontweight="bold")
plt.xlabel("Aplicaci?n")
plt.ylabel("Cantidad de errores")
plt.xticks(rotation=40, ha="right")
plt.tight_layout()
plt.show()

region_errors = (
    df.groupby("macro_region")
      .agg(Total_Peticiones=("is_any_error", "count"),
           Total_Errores=("is_any_error", "sum"))
      .reset_index()
)
region_errors["Tasa de Error (%)"] = (
    region_errors["Total_Errores"] / region_errors["Total_Peticiones"] * 100
).round(2)
region_errors = region_errors.sort_values("Total_Errores", ascending=False)
display(region_errors)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(data=region_errors, x="macro_region", y="Total_Errores",
            ax=axes[0], palette="flare")
axes[0].set_title("Volumen absoluto de errores por regi?n")
axes[0].tick_params(axis="x", rotation=30)
sns.barplot(data=region_errors, x="macro_region", y="Tasa de Error (%)",
            ax=axes[1], palette="crest")
axes[1].set_title("Tasa relativa de error por regi?n")
axes[1].tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Evolucion temporal de errores por hora
if fecha_principal and df[fecha_principal].notna().any():
    time_resampled = (
        df.dropna(subset=[fecha_principal])
          .set_index(fecha_principal)
          .resample("1h")
          .agg(Total_Eventos=("is_any_error", "count"),
               Total_Errores=("is_any_error", "sum"))
          .dropna()
    )
    time_resampled["Tasa_Error_%"] = (
        time_resampled["Total_Errores"] / time_resampled["Total_Eventos"] * 100
    ).round(2)
    plt.figure(figsize=(14, 5))
    plt.plot(time_resampled.index, time_resampled["Total_Errores"],
             color="#d9534f", linewidth=1.5)
    plt.title("Tendencia temporal de errores (agrupado por hora)",
              fontweight="bold")
    plt.xlabel("Fecha y hora (UTC)")
    plt.ylabel("Cantidad de errores por hora")
    plt.tight_layout()
    plt.show()
else:
    print("No se encontraron marcas de tiempo validas para el an?lisis temporal.")
```

## Analisis detallado de eventos `Security`, mensajes e IPs

Se consideran eventos cuyo tipo contiene la palabra completa `SECURITY`. Las IPs, rutas, c?digos HTTP y mensajes sirven para priorizar la investigaci?n; una se?al no confirma por s? sola que exista un ataque.


```python
# Eventos Security y señales para priorizar la investigaci?n
security_mask = df[COL_TIPO].fillna("").astype("string").str.contains(
    r"\bSECURITY\b", case=False, regex=True, na=False
)
df_sec = df.loc[security_mask].copy()

if df_sec.empty:
    print("No se encontraron eventos cuyo tipo contenga SECURITY.")
else:
    df_sec["http_status_num"] = pd.to_numeric(
        df_sec["http_status_code"], errors="coerce"
    )
    message_text = df_sec["sap_function_message"].fillna("").astype(str)
    suspicious_text = message_text.str.contains(
        r"login|auth|unauthori[sz]ed|forbidden|access denied|failed|blocked|"
        r"token|password|injection|brute.?force|anomal|suspicious|attack|exploit",
        case=False, regex=True, na=False
    )
    df_sec["is_suspicious_signal"] = (
        df_sec["http_status_num"].isin([401, 403, 429])
        | (df_sec["http_status_num"] >= 500)
        | suspicious_text
    )
    print("ANaLISIS DE EVENTOS SECURITY")
    print(f"Total de registros: {len(df):,}")
    print(f"Eventos SECURITY: {len(df_sec):,} ({len(df_sec) / len(df) * 100:.2f}%)")
    print(f"Eventos con señales para investigar: "
          f"{int(df_sec['is_suspicious_signal'].sum()):,}")

    print("\nTipos de log encontrados:")
    display(df_sec[COL_TIPO].value_counts(dropna=False).to_frame("Frecuencia"))
    print("\nDistribución por código HTTP:")
    display(df_sec["http_status_num"].value_counts(dropna=False).head(15)
            .to_frame("Frecuencia"))
    print("\nAplicaciones con m?s eventos SECURITY:")
    display(df_sec["sap_function_application"].value_counts().head(10)
            .to_frame("Frecuencia"))

    print("\nIPs con m?s eventos SECURITY:")
    if "client_ip" in df_sec.columns:
        display(df_sec["client_ip"].dropna().astype(str).value_counts().head(15)
                .to_frame("Frecuencia"))
    else:
        print("La columna client_ip no esta disponible.")

    print("\nMensajes SECURITY m?s frecuentes:")
    display(df_sec["sap_function_message"].dropna().astype(str).value_counts()
            .head(15).to_frame("Frecuencia"))

    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    top_apps = df_sec["sap_function_application"].value_counts().head(10)
    sns.barplot(x=top_apps.values, y=top_apps.index, ax=axes[0], color="crimson")
    axes[0].set_title("Aplicaciones afectadas")
    axes[0].set_xlabel("Eventos")

    top_regions = df_sec["macro_region"].value_counts().head(10)
    sns.barplot(x=top_regions.values, y=top_regions.index,
                ax=axes[1], color="darkorange")
    axes[1].set_title("Macro regiones")
    axes[1].set_xlabel("Eventos")

    risk_counts = (
        df_sec["is_suspicious_signal"]
        .map({False: "Sin señal", True: "Con señal"})
        .value_counts()
    )
    sns.barplot(x=risk_counts.values, y=risk_counts.index,
                ax=axes[2], color="purple")
    axes[2].set_title("Señales para investigar")
    axes[2].set_xlabel("Eventos")
    plt.tight_layout()
    plt.show()

    columnas_revision = [
        c for c in ["client_ip", "http_status_num",
                    "headers_http_request_method", "heathers_request_path",
                    "sap_function_message"]
        if c in df_sec.columns
    ]
    eventos_prioritarios = (
        df_sec.loc[df_sec["is_suspicious_signal"], columnas_revision]
        .head(25)
    )
    print("\nMuestra de eventos con señales para investigar:")
    display(eventos_prioritarios)
```
